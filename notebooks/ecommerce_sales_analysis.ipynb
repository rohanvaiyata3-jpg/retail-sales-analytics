```python
Order_by_customer = df.groupby("Customer_ID")["Order_ID"].nunique().sort_values(ascending= False)
print(Order_by_customer)
```

```python
repeat_customer = Order_by_customer[Order_by_customer >1].sum()
print("Repeat_customer:",repeat_customer)
```

```python
repeat_customer_rate = repeat_customer / df["Customer_ID"].nunique() * 100
print("Repeat_customer_rate:", round(repeat_customer_rate),"%")
# print(f"Repeat Customer Rate: {repeat_customer_rate:.2f}%")
```

```python
orders_by_customer = df.groupby("Customer_ID")["Order_ID"].nunique()

repeat_customers = orders_by_customer[orders_by_customer > 1].count()

total_customers = df["Customer_ID"].nunique()

repeat_customer_rate = repeat_customers / total_customers * 100

print("Repeat customers:", repeat_customers)
print("Total customers:", total_customers)
print("Repeat customer rate:", round(repeat_customer_rate, 2), "%")
```

```python
order_frequency = orders_by_customer.value_counts().sort_index()

print(order_frequency)
```

```python
State_Analysis = df.groupby("State")["Net_Sales"].sum().sort_values(ascending= False)
print(State_Analysis)
```

```python
Maharashtra_sales = State_Analysis["Maharashtra"]
Maharashtra_share = Maharashtra_sales / df["Net_Sales"].sum()* 100
print(round(Maharashtra_share, 2),"%")
```

```python
# Orders
order_by_channel = df.groupby("Sales_Channel")["Order_ID"].nunique()

print("Order by channel:")
print(order_by_channel)


# Revenue
channel_sales = df.groupby("Sales_Channel")["Net_Sales"].sum()

print("\nRevenue by channel:")
print(channel_sales)


# AOV
average_order_value = channel_sales / order_by_channel

print("\nAverage Order Value by channel:")
print(round(average_order_value, 2))
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
print(df["Order_Date"].dtype)
```

```python
df["Month"] = df["Order_Date"].dt.to_period("M")
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Month"] = df["Order_Date"].dt.to_period("M")
print(df[["Order_Date", "Month"]].head())
```

```python
monthly_sales = df.groupby("Month")["Net_Sales"].sum().sort_values()
print(monthly_sales)
```

```python
monthly_orders = df.groupby("Month")["Order_ID"].nunique().sort_values()
print(monthly_orders)
```

```python
monthly_analysis = pd.DataFrame({"Revenue": monthly_sales,"Orders": monthly_orders})

monthly_analysis["AOV"] = (monthly_analysis["Revenue"] /monthly_analysis["Orders"])

print(monthly_analysis)
```

```python
df["Year"] = df["Order_Date"].dt.year

yearly_sales = df.groupby("Year")["Net_Sales"].sum()

yearly_orders = df.groupby("Year")["Order_ID"].nunique()

yearly_aov = yearly_sales / yearly_orders

print("Yearly Revenue:")
print(yearly_sales)

print("\nYearly Orders:")
print(yearly_orders)

print("\nYearly AOV:")
print(yearly_aov)
```

```python
revenue_growth = ((yearly_sales[2025] - yearly_sales[2024]) / yearly_sales[2025])*100
print("Revenue YoY Growth:",round(revenue_growth, 2), "%")
```

```python
product_sales = (
    df.groupby("Product")["Net_Sales"]
      .sum()
      .sort_values(ascending=False)
)

print("Top 10 Products by Revenue:")
print(product_sales.head(10))
```

```python
product_quantity = (
    df.groupby("Product")["Quantity"]
      .sum()
      .sort_values(ascending=False)
)

print("\nTop 10 Products by Quantity:")
print(product_quantity.head(10))
```

```python
discount_levels = df["Discount_Percent"].value_counts().sort_index()

print(discount_levels)
```

```python
discount_analysis = (
    df.groupby("Discount_Percent")["Net_Sales"]
      .sum()
      .sort_index()
)

print("Revenue by Discount:")
print(discount_analysis)
```

```python
discount_orders = (
    df.groupby("Discount_Percent")["Order_ID"]
      .nunique()
      .sort_index()
)

print("\nOrders by Discount:")
print(discount_orders)
```

```python
discount_aov = discount_analysis / discount_orders
print("AOV of discount")
print(round(discount_aov, 2))
```
